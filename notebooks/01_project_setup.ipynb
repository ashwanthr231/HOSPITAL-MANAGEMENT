```python
import pandas as pd

df = pd.read_csv("../data/diabetic_data.csv")

print("Dataset loaded successfully!")
print(df.shape)
```

```python
missing = df.isnull().sum()

print(missing)
```

```python
question_marks = (df == "?").sum()

print(question_marks[question_marks > 0])
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100

print(missing_percent[missing_percent > 0])
```

```python
question_mark_percent = (question_marks / len(df)) * 100

print(question_mark_percent[question_mark_percent > 0])
```

```python
duplicates = df.duplicated().sum()

print("Duplicate rows:", duplicates)
```

```python
print(df["gender"].value_counts())
```

```python
print(df["age"].value_counts().sort_index())
```

```python
print(df["readmitted"].value_counts())
```

```python
df_clean = df.copy()

print("Copy created successfully!")
```

```python
df_clean = df_clean.replace("?", pd.NA)

print(" '?' values converted to missing values.")
```

```python
question_marks_after = (df_clean == "?").sum()

print(question_marks_after[question_marks_after > 0])
```

```python
missing_clean = df_clean.isnull().sum()

print(missing_clean[missing_clean > 0])
```

```python
print("Unique encounter IDs:", df_clean["encounter_id"].nunique())
print("Unique patient IDs:", df_clean["patient_nbr"].nunique())
print("Total rows:", len(df_clean))
```

```python
print("Original shape:", df.shape)
print("Cleaned shape :", df_clean.shape)
```

```python
df_clean.to_csv("../data/diabetic_data_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
