```python
import numpy as np

X_train = np.load("../data/processed/X_train.npy")
X_test = np.load("../data/processed/X_test.npy")
y_train = np.load("../data/processed/y_train.npy")
y_test = np.load("../data/processed/y_test.npy")

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Training target distribution:")
print(np.bincount(y_train))

print("\nTesting target distribution:")
print(np.bincount(y_test))
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

print("Model created successfully!")
```

```python
model.fit(X_train, y_train)

print("Model training completed!")
```

```python
y_pred = model.predict(X_test)

print("Predictions completed!")
print("First 20 predictions:")
print(y_pred[:20])
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-Score :", f1)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
from sklearn.metrics import roc_auc_score

y_prob = model.predict_proba(X_test)[:, 1]

roc_auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC:", roc_auc)
```

```python
import numpy as np

X_train = np.load("../data/processed/X_train.npy")
X_test = np.load("../data/processed/X_test.npy")
y_train = np.load("../data/processed/y_train.npy")
y_test = np.load("../data/processed/y_test.npy")

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

print("Random Forest model created successfully!")
```

```python
rf_model.fit(X_train, y_train)

print("Random Forest training completed!")
```

```python
y_pred_rf = rf_model.predict(X_test)

print("Random Forest predictions completed!")
print("First 20 predictions:")
print(y_pred_rf[:20])
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy_rf = accuracy_score(y_test, y_pred_rf)
precision_rf = precision_score(y_test, y_pred_rf)
recall_rf = recall_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf)

print("Random Forest Results")
print("Accuracy :", accuracy_rf)
print("Precision:", precision_rf)
print("Recall   :", recall_rf)
print("F1-Score :", f1_rf)
```

```python
from sklearn.metrics import confusion_matrix

cm_rf = confusion_matrix(y_test, y_pred_rf)

print(cm_rf)
```

```python
y_prob_rf = rf_model.predict_proba(X_test)[:, 1]

print("First 20 probabilities:")
print(y_prob_rf[:20])
```

```python
threshold = 0.30

y_pred_rf_30 = (y_prob_rf >= threshold).astype(int)

print("Predictions with threshold 0.30:")
print(y_pred_rf_30[:20])
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy_rf_30 = accuracy_score(y_test, y_pred_rf_30)
precision_rf_30 = precision_score(y_test, y_pred_rf_30)
recall_rf_30 = recall_score(y_test, y_pred_rf_30)
f1_rf_30 = f1_score(y_test, y_pred_rf_30)

print("Random Forest (Threshold 0.30)")
print("Accuracy :", accuracy_rf_30)
print("Precision:", precision_rf_30)
print("Recall   :", recall_rf_30)
print("F1-Score :", f1_rf_30)
```

```python
import numpy as np
from sklearn.metrics import f1_score

thresholds = np.arange(0.10, 0.51, 0.05)

for threshold in thresholds:
    y_pred_temp = (y_prob_rf >= threshold).astype(int)
    f1 = f1_score(y_test, y_pred_temp)
    recall = recall_score(y_test, y_pred_temp)

    print(
        f"Threshold: {threshold:.2f} | "
        f"F1: {f1:.4f} | "
        f"Recall: {recall:.4f}"
    )
```

```python
best_threshold = 0.25

y_pred_rf_best = (y_prob_rf >= best_threshold).astype(int)

print("Final Random Forest threshold:", best_threshold)
print("Final predictions created!")
```

```python
from sklearn.metrics import confusion_matrix

cm_rf_best = confusion_matrix(y_test, y_pred_rf_best)

print("Final Random Forest Confusion Matrix:")
print(cm_rf_best)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print("Final Random Forest @ 0.25")
print("Accuracy :", accuracy_score(y_test, y_pred_rf_best))
print("Precision:", precision_score(y_test, y_pred_rf_best))
print("Recall   :", recall_score(y_test, y_pred_rf_best))
print("F1-Score :", f1_score(y_test, y_pred_rf_best))
```
